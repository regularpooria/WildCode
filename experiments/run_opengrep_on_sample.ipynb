```python
import os
import uuid
from pathlib import Path
from typing import Dict, List


dir_path = Path(f"../files/temp/")
dir_path.mkdir(parents=True, exist_ok=True)
```

```python
def save_code_snippet(code: str, language: str) -> str:
    """
    Save a code snippet to a file in files/temp/{language}/ directory.
    
    Args:
        code: The code snippet to save
        language: The programming language (e.g., 'python', 'javascript', 'java')
    
    Returns:
        str: The UUID4 filename (without extension) used to save the file
    """
    # Create the directory if it doesn't exist
    
    # Generate a UUID4 for the filename
    file_uuid = str(uuid.uuid4())
    
    # Determine file extension based on language
    extensions = {
        'python': '.py',
        'javascript': '.js',
        'java': '.java',
        'c': '.c',
        'cpp': '.cpp',
        'csharp': '.cs',
        'php': '.php',
        'ruby': '.rb',
        'go': '.go',
        'rust': '.rs',
        'swift': '.swift',
        'kotlin': '.kt',
    }
    
    extension = extensions.get(language.lower(), '.txt')
    file_path = dir_path / f"{file_uuid}{extension}"
    
    # Write the code to the file
    with open(file_path, 'w', encoding='utf-8') as f:
        f.write(code)
    
    return file_uuid


def save_code_snippets_batch(code_snippets: List[str], language: str) -> Dict[int, str]:
    """
    Save multiple code snippets to files and return a mapping of input index to UUID.
    
    Args:
        code_snippets: List of code snippets to save
        language: The programming language for all snippets
    
    Returns:
        Dict[int, str]: Mapping of input list index to UUID filename
    """
    mapping = {}
    
    for idx, code in enumerate(code_snippets):
        file_uuid = save_code_snippet(code, language)
        mapping[idx] = file_uuid
    
    return mapping
```

```python

# Example usage:
if __name__ == "__main__":
    # Test with a single snippet
    sample_code = """def hello_world():
    print("Hello, World!")"""
    
    uuid_result = save_code_snippet(sample_code, "python")
    print(f"Saved single snippet with UUID: {uuid_result}")
    
    # Test with multiple snippets
    snippets = [
        "def add(a, b): return a + b",
        "def subtract(a, b): return a - b",
        "def multiply(a, b): return a * b"
    ]
    
    mapping = save_code_snippets_batch(snippets, "python")
    print(f"\nSaved {len(snippets)} snippets:")
    for idx, uuid_val in mapping.items():
        print(f"  Index {idx} -> {uuid_val}")
```
