```python
from transformers import pipeline
```

```python
from datasets import load_dataset

# Login using e.g. `huggingface-cli login` to access this dataset
ds = load_dataset("regularpooria/wildcode")
```

```python
classifier = pipeline("zero-shot-classification",
                      model="facebook/bart-large-mnli")
```

```python
labels = {
    "beginner": "print statements, variables, loops, functions, classes, syntax",
    "intermediate": "frameworks, debugging, APIs, libraries, data processing, workflow",
    "expert": "algorithms, neural networks, optimization, system design, research"
}
```

```python
from tqdm.auto import tqdm
import math

def classify_text(texts, batch_size=32, multi_class=False):
    """
    Classify one or many texts using the global `classifier`
    and global `labels` dict (label → description).
    """

    # Accept single string
    if isinstance(texts, str):
        texts = [texts]

    label_names = list(labels.keys())
    label_descs = list(labels.values())

    results = []
    num_batches = math.ceil(len(texts) / batch_size)

    for i in tqdm(range(num_batches), desc="Classifying"):
        batch = texts[i * batch_size : (i + 1) * batch_size]

        out = classifier(
            batch,
            candidate_labels=label_descs,
            hypothesis_template="This question is about {}."

        )

        if isinstance(out, dict):
            out = [out]

        # Map back from description → label key
        for r in out:
            mapped_labels = []
            for desc_label in r["labels"]:
                key = label_names[label_descs.index(desc_label)]
                mapped_labels.append(key)

            r["labels"] = mapped_labels

        results.extend(out)

    return results
```

```python
classify_text("how do i print hello world")
```

```python
# Beginner examples
beginner_examples = [
    "How do I print 'Hello World' in Python?",
    "What is the difference between a list and a tuple?",
    "How do I write a for loop that counts from 1 to 10?",
    "How do I define a function that adds two numbers?",
    "How do I create a class in Python?",
    "What is the syntax for an if-else statement?",
    "How do I assign a value to a variable?",
    "How do I concatenate two strings?",
    "How do I use the len() function?",
    "How do I import a module in Python?"
]

# Intermediate examples
intermediate_examples = [
    "How do I use the requests library to fetch data from an API?",
    "How do I debug a function that returns the wrong output?",
    "How can I read a CSV file using pandas?",
    "How do I combine multiple lists into a dictionary?",
    "How do I handle exceptions when opening a file?",
    "How do I create a virtual environment and install packages?",
    "How do I filter a list of numbers using a lambda function?",
    "How can I merge two dataframes in pandas?",
    "How do I authenticate with an API using a token?",
    "How do I write unit tests for a function?"
]

# Expert examples
expert_examples = [
    "How do I implement a convolutional neural network from scratch in PyTorch?",
    "How do I optimize a sorting algorithm to handle millions of elements?",
    "How do I design a scalable distributed system for a web application?",
    "How do I implement backpropagation in a custom neural network?",
    "How do I reduce memory usage in a deep learning model?",
    "How do I implement reinforcement learning for a game AI?",
    "How do I design a recommendation system with collaborative filtering?",
    "How do I apply gradient clipping to stabilize training in deep networks?",
    "How do I analyze and optimize the time complexity of a recursive algorithm?",
    "How do I design a caching strategy for a high-throughput database?"
]

# Combine all examples
all_examples = beginner_examples + intermediate_examples + expert_examples

# Example usage with your classifier
results = classify_text(all_examples)

# Print results
for example, res in zip(all_examples, results):
    print(f"Question: {example}")
    for label, score in zip(res["labels"], res["scores"]):
        print(f"  {label}: {score:.1f}%")
    print()
```
