```python
import json
import pandas
```

```python
data = []
```

```python
with open("../../results/zeroshot_results_user_expertise.jsonl", "r", encoding="utf-8") as f:
    for line in f.readlines():
        data.append(json.loads(line))
```

```python
import pickle
with open("../../results/zeroshot_results_user_expertise.pkl", "wb") as f:
    pickle.dump(data, f)
```

```python
threshhold = 0.6
```

```python
filtered = []
for row in data:
    if max(row["scores"].values()) >= threshhold:
        filtered.append(row)
```

```python
len(filtered)
```

```python
output = {"beginner": 0, "intermediate": 0, "expert": 0}
for row in filtered:
    output[list(row["scores"].keys())[0]] += 1 
    
print(output)
```

```python
vulnerabilities = pandas.read_csv("../../results/opengrep_results.csv")
```

```python
for i in range(len(filtered)):
    filtered[i]["vulnerabilities"] = None
for index, row in enumerate(filtered):
    if row["conversation_hash"] in vulnerabilities["conversation_hash"].tolist():
        filtered[index]["vulnerabilities"] = vulnerabilities[vulnerabilities["conversation_hash"] == row["conversation_hash"]].to_dict(orient="records")
```

Determining how many vulnerabilities there are out of the classifications that have passed the threshold.

```python
output = {"beginner": 0, "intermediate": 0, "expert": 0}
for row in filtered:
    if row["vulnerabilities"] is None:
        continue
    output[list(row["scores"].keys())[0]] += 1 
print(output)
```
