```python
import json
```

```python
data = []
```

```python
with open("./zero_shot_classification_results.jsonl", "r", encoding="utf-8") as f:
    for line in f.readlines():
        data.append(json.loads(line))
```

```python
threshhold = 0.6
```

```python
filtered = []
for row in data:
    if max(row["scores"].values()) > threshhold:
        filtered.append(row)
```

```python
len(filtered)
```

```python
output = {"beginner": 0, "intermediate": 0, "expert": 0}
for row in filtered:
    output[list(row["scores"].keys())[0]] += 1 
```
